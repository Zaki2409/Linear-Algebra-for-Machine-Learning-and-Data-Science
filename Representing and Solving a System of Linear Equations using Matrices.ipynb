```python
import numpy as np
np.version.version
```

```python
A = np.array([
    [4, -3, 1],
    [2, 1, 3],
    [-1, 2, -5]
],dtype=np.dtype(float))

b = np.array([10, 0, 17], dtype=np.dtype(float))
print("Matrix A:")
print(A)
print("\n Array b:")
print(b)
```

```python
print(f"Shape of A: {np.shape(A)}")
print(f"Shape of B: {np.shape(b)}")
```

```python
det = np.linalg.det(A)
print(f"Determinant of matrix A: {det:.2f}")
```

```python
print(np.linalg.solve(A, b))
```
